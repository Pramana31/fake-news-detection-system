```python
#do feature selection
news=news.drop(['title'],axis=1)
```

```python
news=news.sample(frac=1)                    #reshuffle 
news.reset_index(inplace=True)
news=news.drop(['index'],axis=1)
```

```python
news
```

```python
#import regular expression module 
import re
```

```python
def wordopt(text):
    #convert into lowercase 
    text=text.lower()

    #remove URLs
    text=re.sub(r'https?://S+|www\.\S+','',text)

    #remove HTML tags
    text=re.sub(r'<.*?>','',text)

    #remove punctuation
    text=re.sub(r'[^\w\s]','',text)

    #remove digits
    text=re.sub(r'\d','',text)

    #remove newline charcters
    text=re.sub(r'\n',' ',text)

    return text
```

```python
news['text']=news['text'].apply(wordopt)
```

```python
news['text']
```

```python
x=news['text']
y=news['label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorization=TfidfVectorizer()
```

```python
xv_train=vectorization.fit_transform(x_train)
```

```python
xv_test=vectorization.transform(x_test)
```

```python
xv_train
```

```python
xv_test
```

```python
#build logistic regression model
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
lr.fit(xv_train,y_train)
```

```python
pred_lr=lr.predict(xv_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
print(classification_report(y_test, pred_lr))
print('Accuracy:',100*accuracy_score(y_test,pred_lr))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
```

```python
dt.fit(xv_train,y_train)
```

```python
pred_dt=dt.predict(xv_test)
```

```python
print(classification_report(y_test, pred_dt))
print('Accuracy:',100*accuracy_score(y_test,pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
```

```python
rf.fit(xv_train,y_train)
```

```python
pred_rf=rf.predict(xv_test)
```

```python
print(classification_report(y_test, pred_rf))
print('Accuracy:',100*accuracy_score(y_test,pred_rf))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gb=GradientBoostingClassifier()
```

```python
gb.fit(xv_train,y_train)
```

```python
pred_gb=gb.predict(xv_test)
```

```python
print(classification_report(y_test, pred_gb))
print('Accuracy:',100*accuracy_score(y_test,pred_gb))
```

```python
def output_label(n):
    if n=='FAKE':
        return "It is a fake news"
    elif n=='REAL':
        return "It is a genuine news"
```

```python
def manual_testing(news):
    testing_news = {"text": [news]}

    new_def_test = pd.DataFrame(testing_news)
    new_def_test["text"] = new_def_test["text"].apply(wordopt)

    new_x_test = new_def_test["text"]

    # Assuming vectorization is your trained vectorizer
    new_xv_test = vectorization.transform(new_x_test)

    pred_lr = lr.predict(new_xv_test)
    pred_dt = dt.predict(new_xv_test)
    pred_gb = gb.predict(new_xv_test)
    pred_rf = rf.predict(new_xv_test)

    return (
        "\nLR Prediction: {}"
        "\nDT Prediction: {}"
        "\nGB Prediction: {}"
        "\nRF Prediction: {}"
    ).format(
        output_label(pred_lr),
        output_label(pred_dt),
        output_label(pred_gb),
        output_label(pred_rf)
    )
```

```python
news = input("Enter news: ")
print(manual_testing(news))
```
